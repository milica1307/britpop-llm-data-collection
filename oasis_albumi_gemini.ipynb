In accordance with the established protocol for technical tasks—which requires strictly verified data from original sources and prohibits the use of generative predictions or approximations—I will not provide estimated parameters or unverified values for this track's audio features or metadata."

LLM ne može direktno da pristupi ovim podacima. Podatke može da dobije na dva načina:
1. Koristi svoje trening podatke
2. Simulira API tako što koristi svoje znanje o pesmama da rekonstruiše vrednosti koje bi Spotify Web API potencijalno vratio.

Ni jedan od ovih načina nam ne daje sigurno tačne podatke. LLM se oslanja na verovatnoću i na taj način dolazi do ovih podataka. Takođe, iako se u promptu naglasi koja verzija albuma treba da se koristi, on često može da vrati podatke koje pamti za neku drugu verziju.

```python
definitely_maybe2=pd.read_csv("definitely_maybe_oasis2.csv")
definitely_maybe2.head()
```

```python
# Definisanje numeričkih kolona koje poredimo
features = [
    'duration_ms', 'acousticness', 'danceability', 'instrumentalness',
    'liveness', 'loudness', 'speechiness', 'tempo', 'valence'
]

# Postavljanje 'id' kao indeksa (ključ za uparivanje)
df1_indexed = definitely_maybe.set_index('id')[features]
df2_indexed = definitely_maybe2.set_index('id')[features]

# Izračunavanje apsolutne razlike
diff_df = (df1_indexed - df2_indexed).abs()

# Vraćanje naziva pesama radi lakše čitljivosti
names_map = definitely_maybe.set_index('id')['track_name']
diff_df.insert(0, 'track_name', diff_df.index.map(names_map))

# Čuvanje rezultata
diff_df.to_csv('razlike_u_datasetovima.csv')
```

### Poređenje
Kada se uporede vrednosti ova dva dataset-a vidimo da kod većine podataka postoji neko odstupanje. Največa odstupanja sam primetila kod feature-a loudness i tempo, ali i kod ostalih se dešava da vrednost značajno odstupa.

### Provera konzistentnosti podataka dobijenih od LLM-a

Poredjenje dva pokusaja (identican prompt poslat dva puta).

```python
features = ['duration_ms', 'acousticness', 'danceability', 'instrumentalness',
            'liveness', 'loudness', 'speechiness', 'tempo', 'valence']

df1 = definitely_maybe.copy()
df1.columns = df1.columns.str.strip()
df1['track_name_key'] = df1['track_name'].str.strip().str.lower()

df2 = definitely_maybe2.copy()
df2.columns = df2.columns.str.strip()
df2['track_name_key'] = df2['track_name'].str.strip().str.lower()

d1_indexed = df1.set_index('track_name_key')[features]
d2_indexed = df2.set_index('track_name_key')[features]

# Prosecno i maksimalno apsolutno odstupanje: Pokusaj 1 vs Pokusaj 2
odstupanje_prosek = (d1_indexed - d2_indexed).abs().mean()
odstupanje_max = (d1_indexed - d2_indexed).abs().max()

tabela_odstupanja = pd.DataFrame({
    'Prosecno apsolutno odstupanje': odstupanje_prosek,
    'Maksimalno apsolutno odstupanje': odstupanje_max
}).round(3)

tabela_odstupanja.to_csv('tabela_1_odstupanja_llm.csv')
tabela_odstupanja
```

```python
# Relativno odstupanje u odnosu na ukupan opseg vrednosti svake karakteristike
opseg_vrednosti = {
    'duration_ms': 400000,
    'tempo': 200,
    'loudness': 20,
    'valence': 1, 'liveness': 1, 'instrumentalness': 1,
    'danceability': 1, 'acousticness': 1, 'speechiness': 1
}

relativno_odstupanje = pd.Series({
    f: odstupanje_prosek[f] / opseg_vrednosti[f] * 100 for f in features
}).round(1)

relativno_odstupanje.name = 'Relativno odstupanje (%)'
relativno_odstupanje.to_csv('relativno_odstupanje_llm.csv')
relativno_odstupanje
```
