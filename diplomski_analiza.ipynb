### Spotify API alternativa
Na sledećem linku se nalazi alternativa API-ja koja može da se koristi za dobijanje audio features sa Spotify-a: https://rapidapi.com/musicae-musicae-default/api/spotify-extended-audio-features-api. Registracijom na platformu se dobija API ključ uz koji se dobija ograničen broj requestova mesečno.

Korišćen je LLM kome je poslat sledeći prompt: Potrebno je koriščenjem Spotify Extended Audio Features API prikupiti feature sa Spotify-a za pesmu Live Forever - Oasis. API overview se nalazi na sledećem linku: https://rapidapi.com/musicae-musicae-default/api/spotify-extended-audio-features-api. Spotify pesma se nalazi na sledećem linku: https://open.spotify.com/track/1KgfeuVn5OlsBEtoEmBa1t?si=bfc526331f0a4ef9. Podaci koji treba da budu prikupljeni za datu pesmu su: track_name, artist, album, duration_ms, release_date, acousticness, danceability, instrumentalness, liveness, loudness, mode, speechiness, tempo, time_signature, valence. Potrebno je napisati Python skriptu koja će da vrši scraping ovih podataka i da na kraju pretvori u .csv fajl gde će kolone predstavljati prikupljene feature.

Uz minimalne izmene koda koji je LLM generisao dobijen je kod ispod i na taj način možemo da dobijemo feature. Featuri koje smo dobili na ovaj način se malo razlikuju u vrednostima od onih koje nam je generisao LLM ranije.

```python
import requests
import pandas as pd

# Konfiguracija
track_id = "1KgfeuVn5OlsBEtoEmBa1t"  # Live Forever ID
api_key = "KLJUC"
host = "spotify-extended-audio-features-api.p.rapidapi.com"

headers = {
    "X-RapidAPI-Key": api_key,
    "X-RapidAPI-Host": host
}

def get_full_track_data(t_id):
    try:
        # 1. Poziv za osnovne podatke (GET track)
        track_url = f"https://{host}/v1/tracks/{t_id}"
        t_res = requests.get(track_url, headers=headers)
        t_res.raise_for_status()
        t_data = t_res.json()

        # 2. Poziv za audio features (GET track audio features)
        features_url = f"https://{host}/v1/audio-features/{t_id}"
        f_res = requests.get(features_url, headers=headers)
        f_res.raise_for_status()
        f_data = f_res.json()

        # Spajanje podataka u jedan rečnik
        combined_data = {
            # Iz prvog API poziva
            "track_name": t_data.get("name"),
            "artist": ", ".join([a["name"] for a in t_data.get("artists", [])]),
            "album": t_data.get("album", {}).get("name"),
            "release_date": t_data.get("album", {}).get("release_date"),
            "duration_ms": t_data.get("duration_ms"),

            # Iz drugog API poziva
            "acousticness": f_data.get("acousticness"),
            "danceability": f_data.get("danceability"),
            "instrumentalness": f_data.get("instrumentalness"),
            "liveness": f_data.get("liveness"),
            "loudness": f_data.get("loudness"),
            "mode": f_data.get("mode"),
            "speechiness": f_data.get("speechiness"),
            "tempo": f_data.get("tempo"),
            "time_signature": f_data.get("time_signature"),
            "valence": f_data.get("valence")
        }

        # Kreiranje CSV-a
        df = pd.DataFrame([combined_data])
        df.to_csv("oasis_live_forever_full_features.csv", index=False, encoding='utf-8')

        print("Podaci uspešno prikupljeni i sačuvani!")
        return df

    except requests.exceptions.HTTPError as err:
        print(f"HTTP greška: {err}")
    except Exception as e:
        print(f"Došlo je do greške: {e}")

if __name__ == "__main__":
    result_df = get_full_track_data(track_id)
```
